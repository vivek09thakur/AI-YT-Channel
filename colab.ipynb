```python
#@title Setup
import os, sys, subprocess
REPO_URL = "https://github.com/vivek09thakur/AI-YT-Channel.git" #@param {type:"string"}
BRANCH = "main" #@param {type:"string"}
REPO_DIR = "/content/AI-YT-Channel"
if not os.path.isdir(REPO_DIR):
    subprocess.run(["git", "clone", "-q", "-b", BRANCH, REPO_URL, REPO_DIR], check=True)
else:
    subprocess.run(["git", "-C", REPO_DIR, "pull", "-q"], check=False)
os.chdir(REPO_DIR)
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
sys.path.insert(0, REPO_DIR)
print("ready")
```

```python
#@title API Keys (Colab Secrets override these fields)
import os
GEMINI_API_KEY = "" #@param {type:"string"}
PEXELS_API_KEY = "" #@param {type:"string"}
ASSEMBLY_AI_API_KEY = "" #@param {type:"string"}
GEMINI_MODEL = "gemini-2.5-flash" #@param {type:"string"}

def _secret(name: str, fallback: str) -> str:
    try:
        from google.colab import userdata
        return userdata.get(name) or fallback
    except Exception:
        return fallback

os.environ.update({
    "GEMINI_MODEL": GEMINI_MODEL,
    "GEMINI_API_KEY": _secret("GEMINI_API_KEY", GEMINI_API_KEY),
    "PEXELS_API_KEY": _secret("PEXELS_API_KEY", PEXELS_API_KEY),
    "ASSEMBLY_AI_API_KEY": _secret("ASSEMBLY_AI_API_KEY", ASSEMBLY_AI_API_KEY),
})
print({k: bool(os.environ.get(k)) for k in ("GEMINI_API_KEY", "PEXELS_API_KEY", "ASSEMBLY_AI_API_KEY")})
```

```python
#@title Generate
import os
from engine import Options, run
SUBJECT = "" #@param {type:"string"}
PARAGRAPHS = 1 #@param {type:"integer"}
VOICE = "en_us_001" #@param ["en_us_001", "en_us_002", "en_us_006", "en_us_007", "en_us_009", "en_us_010", "en_uk_001", "en_uk_003", "en_au_001", "en_au_002", "en_male_narration", "en_male_funny", "en_female_emotional", "fr_001", "fr_002", "de_001", "de_002", "es_002", "es_mx_002", "br_001", "br_003", "br_004", "br_005", "id_001", "jp_001", "jp_003", "jp_005", "jp_006", "kr_002", "kr_003", "kr_004"]
POSITION = "center,bottom" #@param ["center,bottom", "center,top", "center,center", "left,bottom", "right,bottom"]
COLOR = "#FFFF00" #@param ["#FFFF00", "#f4a261", "#e63946", "#1d3557", "#ffffff", "#03071e"]
CUSTOM_PROMPT = "" #@param {type:"string"}
UPLOAD_TO_YOUTUBE = False #@param {type:"boolean"}

OUTPUT = run(Options(
    subject=SUBJECT,
    paragraphs=PARAGRAPHS,
    voice=VOICE,
    position=POSITION,
    color=COLOR,
    custom_prompt=CUSTOM_PROMPT,
    threads=os.cpu_count() or 2,
    upload=UPLOAD_TO_YOUTUBE,
))
print(OUTPUT)
```
